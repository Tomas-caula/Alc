## 1. Lectura de datos

Implementar la función `cargarDataset(carpeta)` que lea las matrices de embeddings 
precalculadas (en formato .npy) de las carpetas `train` y `val`, y genere:

- $X_t$, $Y_t$ → entrenamiento
- $X_v$, $Y_v$ → validación

Cada columna de $X$ corresponde a un embedding.
Cada columna de $Y$ es un vector $[1,0]^T$ (gato) o $[0,1]^T$ (perro).


```python
import numpy as np
from moduloALC import *

# Cargamos los datos para ver e interpretar su forma
carpetaDatos = r'template-alumnos\cats_and_dogs'
Xt, Yt, Xv, Yv = cargarDataset(carpetaDatos)
print("X_training:", Xt.shape, "Y_training:", Yt.shape)
print("Xv:", Xv.shape, "Yv:", Yv.shape)
print("Etiquetas únicas:", np.unique(Yt), np.unique(Yv))
```

## 2. Cálculo de $W$ usando distintos métodos

A partir de $X_t$, $Y_t$:
1. Ecuaciones normales (Cholesky)
2. SVD
3. QR (Householder y Gram-Schmidt)

Cada método debe calcular la pseudo-inversa X⁺ correspondiente y luego $W = Y_t X⁺$

```python
XT = Xt.T
Q, R= QR_con_GS(XT, False)

WGS = pinvGramSchmidt(Q, R, Yt)
```

```python
XT = Xt.T
Q, R= QR_con_HH(XT)

WHH = pinvHouseHolder(Q, R, Yt)
```

```python
A = Xt @ traspuesta(Xt)
L = calculaCholesky(A)                                 

WEN = pinvEcuacionesNormales(L, Yt, traspuesta(Xt))
```

```python
U, s, VT = svd_reducida(Xt)

WSVD = pinvSVD(U, s, VT, Yt, tol=1e-12)
```

```python
# HH 
YHH = WHH @ Xv

pred = argmax_axis0(YHH)
true = argmax_axis0(Yv)

MHH, acc, p = confusion_matrix(true, pred)
print(MHH)
print("Accuracy:", acc)
print('Precisión:', p)
```

```python
# GS
YGS = WGS @ Xv

pred = argmax_axis0(YGS)
true = argmax_axis0(Yv)

MGS, acc, p = confusion_matrix(true, pred)
print(MGS)
print("Accuracy:", acc)
print('Precisión:', p)
```

```python
# SVD
YSVD = WSVD @ Xv

pred = argmax_axis0(YSVD)
true = argmax_axis0(Yv)

MSVD, acc, p = confusion_matrix(true, pred)
print(MSVD)
print("Accuracy:", acc)
print('Precisión:', p)
```

```python
YEN = WEN @ Xv

pred = argmax_axis0(YEN)
true = argmax_axis0(Yv)

MEN, acc, p = confusion_matrix(true, pred)
print(MEN)
print("Accuracy:", acc)
print('Precisión:', p)
```
